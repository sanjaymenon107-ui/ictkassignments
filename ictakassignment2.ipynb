```python
#numpy array operations
import numpy as np

a = np.random.randint(1, 51, (5, 5))

print("Array:")
print(a)

sums = a.sum(axis=1)
print("Row with maximum sum:", a[np.argmax(sums)])

mean = a.mean()
a[a > mean] = 0

print("Mean:", mean)
print("New array:")
print(a)
```

```python
#data virtualization matplotlib
import matplotlib.pyplot as plt

plt.hist(data["sepal_length"], bins=15)
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()

for s in data["species"].unique():
    x = data[data["species"] == s]
    plt.scatter(x["sepal_length"], x["petal_length"], label=s)

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.legend()
plt.show()
```

```python
#seaborn
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(data)
plt.show()

sns.boxplot(x="species", y="petal_width", data=data)
plt.show()
```
