```python
#Simple ATM Calculator
balance = int(input("Enter balance: ")) #enter balance
withdrawal = int(input("Enter withdrawal amount: ")) #told user to enter withdrawal amount

if withdrawal <= balance:
    balance = balance - withdrawal #code to get the balance
    print("Transaction successful. Remaining balance:", balance) #output code for showing balance
else:
    print("Error: Insufficient balance.") #output code for shwoing insufficient balance
```

```python
#electricity bill generator
units = int(input("Enter units consumed: ")) #to enter the units consumed
if units <= 100:
    bill = units * 5 #to calculate 100 units

elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7) #to calculate bill till 200

else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10) #to calculate after 200

print("Bill amount = ₹", bill) #to show the bill
```

```python
#temperature converter
temp = float(input("Enter temperature: "))
unit = input("Enter unit (C or F): ")  # to get the temp and unit

if unit == "C":
    fahrenheit = (temp * 9/5) + 32
    print("Temperature in Fahrenheit:", fahrenheit, "°F")  # to convert celsius to fahrenheit

elif unit == "F":
    celsius = (temp - 32) * 5/9
    print("Temperature in Celsius:", celsius, "°C")  # to convert fahrenheit to celsius

else:
    print("Invalid unit. Please enter 'C' or 'F'.")  # display error for invalid unit
```

```python
#MOBILE RECHARGE VALIDITY
from datetime import datetime
recharge_date = input("Enter recharge date (DD-MM-YYYY): ") #tells to enter the date user recharged
today_date = input("Enter today's date (DD-MM-YYYY): ") #enter todays date
recharge_date = datetime.strptime(recharge_date, "%d-%m-%Y") #TO CONVERT RECHARGE DATE TO DATE FORMAT
today_date = datetime.strptime(today_date, "%d-%m-%Y") #CONVERT TODAYS DATE INTO DATE FORMAT
days = (today_date - recharge_date).days #FIND NUMBER OF DAYS BETWEEN DATES
if days <= 28:
    print("Plan is still valid.")  #CHECKING WHETHER PLAN IS VALID
else:
    print("Plan expired. Recharge again.") #SHOWING THE ERROR MESSAGE THAT PLAN IS EXPIRED
```

```python
#SUM OF EVEN NUMBERS
num = int(input("Enter a number: "))
sum_even = 0 #INITIALIZE THE SUM
for i in range(1, num + 1):
    if i % 2 == 0: #CHECK THE NUMBER IS EVEN OR NOT
     sum_even = sum_even + i #ADD THE EVEN NO TO SUM
print("Sum of even numbers:", sum_even)

```

```python
#DICTIONARY-STUDENT MARKS SUMMARY
students = {}

for i in range(3):
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))
    students[name] = marks #STORE THE NAME AND MARKS IN DICTIONARY
average = sum(students.values()) / 3 #CALCULATE AVERAGE OF 3 STUDENTS MARKS
print("Student marks:", students) #DISPLAY STUDENT MARKS AND AVERAGE MARKS
print("Average marks:", average)
```

```python
#DICTIONARY WORD FREQUENCY
sentence = input("Enter a sentence: ")

words = sentence.split()
frequency = {}   #CREATE EMPTY DICTIONARY

for word in words: #CHECK EACH WORD IN SENTENCE
    if word in frequency:
        frequency[word] = frequency[word] + 1 #INCREASE FREQUENCY OF WORD
    else:
        frequency[word] = 1

for word in frequency:
    print(word + ":", frequency[word]) #DISPLAY FREQUENCY OF EACH WORD
```

```python
#FUNCTIONS-CHECK PRIME
def is_prime(num): #DEFINE THE FUNCTION
    if num < 2: #CHECK IF NUMNER IS LESS THAN 2
        return False
    for i in range(2, num): #CHECK IF THE NUMBER IS DIVISIBLE BY ANY NUMBER
        if num % i == 0: #CHECK IF THE NUMBER IS DIVISIBLE
            return False
    return True #RETURN TRUE IF NUMBER IS PRIME
num = int(input("Enter a number: ")) #GET A NUMBER FROM USER
if is_prime(num):     #CHECK NUMBER IS PRIME OR NOT
    print("The number is prime.") #DISPLAY WHETHER GIVEN NO IS PRIME OR NOT
else:
    print("The number is not prime.")
```

```python
#VOWEL COUNTER

def count_vowels(text):
    count = 0
    for letter in text: #check the letters in string
        if letter.lower() in "aeiou": #check vowel in the letter
            count = count + 1 #increase the vowel count by 1
    return count
text = input("Enter a string: ") #get the entered string
print("Number of vowels:", count_vowels(text)) #display no of vowels

```

```python
#COUNT DIVISIBLE NUMBERS
n = int(input("Enter range limit N: ")) #to get the range and divisor
d = int(input("Enter divisor D: "))
count = 0
for i in range(1, n + 1): #to check numbers from 1 to n
    if i % d == 0:        #check if number is divisible by D
        count = count + 1
print("Count of numbers divisible by", d, ":", count) # showw the numbers divisible by d
```
